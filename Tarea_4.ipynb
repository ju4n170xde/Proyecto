```python
"Área y volumen del dodecaedro e icosaedro"
"Esta tarea me ayudó a conocer las formulas de estas dos figuras (dodecaedro e icosaedro), con sus clases y las del profe Ledesma me ayudaron a poder realizar la mayoría de mi tarea, sin embargo, tuve que buscar unos comandos en internet porque no sabía como realizarlos "

import math

def calcular_solidos():
    print("--- Calculadora de Sólidos Platónicos ---")
    a = float(input("Ingresa la longitud de la arista (a): "))

    # Cálculos Dodecaedro
    area_dode = 3 * math.sqrt(25 + 10 * math.sqrt(5)) * a**2
    vol_dode = ((15 + 7 * math.sqrt(5)) / 4) * a**3

    # Cálculos Icosaedro
    area_ico = 5 * math.sqrt(3) * a**2
    vol_ico = (5 * (3 + math.sqrt(5)) / 12) * a**3

    print(f"\nResultados para un Dodecaedro:")
    print(f"- Área: {area_dode:.2f}")
    print(f"- Volumen: {vol_dode:.2f}")

    print(f"\nResultados para un Icosaedro:")
    print(f"- Área: {area_ico:.2f}")
    print(f"- Volumen: {vol_ico:.2f}")

calcular_solidos()
```
